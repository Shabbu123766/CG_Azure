```
from google.colab import userdata
from openai import OpenAI
import gradio as gr

# Load API key
api_key = userdata.get("open_api_key")

# Azure details
my_endpoint = "https://shabbu456.services.ai.azure.com/api/projects/proj-default"
my_agent = "AI-Customer-Chatbot"
my_version = "2"

# Client
client = OpenAI(
    base_url=f"{my_endpoint}/openai/v1",
    api_key=api_key
)

# System Prompt (Guardrails)
system_prompt = """
You are an AI Customer Support Assistant.

Rules:
- Only answer customer support queries (orders, refunds, delivery, account issues).
- Do NOT answer unrelated questions.
- Do NOT generate harmful or illegal content.
- If the query is outside your scope, say: "I can only help with customer support queries."
- If unsure, say: "Please contact support team at support@company.com."
- Always respond politely and professionally.
"""

# Chat function
def chatbot(user_input, customer_id):
    try:
        response = client.responses.create(
            input=[
                {
                    "type": "message",
                    "role": "system",
                    "content": [{"type": "input_text", "text": system_prompt}]
                },
                {
                    "type": "message",
                    "role": "user",
                    "content": [
                        {
                            "type": "input_text",
                            "text": f"Customer ID: {customer_id}\nQuery: {user_input}"
                        }
                    ]
                }
            ],
            extra_body={
                "agent_reference": {
                    "name": my_agent,
                    "version": my_version,
                    "type": "agent_reference"
                }
            },
        )

        return f"[Customer ID: {customer_id}]\n{response.output_text}"

    except Exception as e:
        return f"Error: {str(e)}"


# Gradio UI
interface = gr.Interface(
    fn=chatbot,
    inputs=[
        gr.Textbox(label="Enter your question"),
        gr.Textbox(label="Customer ID")
    ],
    outputs="text",
    title="AI Customer Support Chatbot",
    description="Ask about orders, refunds, delivery, etc."
)

interface.launch()
```

```
from google.colab import userdata
from openai import OpenAI

# Load API Key
api_key = userdata.get("open_api_key")

# Azure details
endpoint = "https://shabbu456.services.ai.azure.com/api/projects/proj-default"
agent_name = "AI-Interview-Simulator-with-Feedback"
agent_version = "1"

client = OpenAI(
    base_url=f"{endpoint}/openai/v1",
    api_key=api_key
)

# Function to talk to agent
def ask_agent(user_input):
    response = client.responses.create(
        input=[
            {
                "type": "message",
                "role": "user",
                "content": [{"type": "input_text", "text": user_input}]
            }
        ],
        extra_body={
            "agent_reference": {
                "name": agent_name,
                "version": agent_version,
                "type": "agent_reference"
            }
        },
    )
    return response.output_text



print(" AI Interview Simulator\n")

while True:
    user_input = input("You: ")

    if user_input.lower() in ["exit", "quit"]:
        print("Exiting...")
        break

    reply = ask_agent(user_input)
    print("\nAI:\n", reply)
```
